```python
from collections import deque
import random

def trace_infection(graph, start, max_depth=2):
    if start not in graph:
        print(f"시작 노드 '{start}'가 그래프에 없습니다.")
        return

    visited = {start}
    queue = deque([(start, 0, [start])])
    infected_list = []

    while queue:
        current, depth, path = queue.popleft()

        if 1 <= depth <= max_depth:
            infected_list.append((current, depth, path))

        if depth < max_depth:
            for neighbor in graph.get(current, []):
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append((neighbor, depth + 1, path + [neighbor]))

    print(f"=== [최초 감염자: {start} / 최대 탐색 깊이(max_depth): {max_depth}] ===")
    print(f"총 노드 수: {len(graph)}개 | 감염된 노드 수(본인 제외): {len(infected_list)}명")
    for node, level, path in infected_list[:15]:
        route = " -> ".join(path) if len(path) <= 6 else f"{path[0]} -> ... -> {path[-1]}"
        print(f"- {node} | 거리(레벨): {level}단계 | 경로: {route}")
    if len(infected_list) > 15:
        print(f"... 외 {len(infected_list) - 15}명 생략")
    print()

def generate_random_graph(num_nodes, max_edges=3, seed=42):
    random.seed(seed)
    nodes = [f"Node_{i}" for i in range(num_nodes)]
    g = {node: [] for node in nodes}
    for node in nodes:
        k = random.randint(1, max_edges)
        candidates = [n for n in nodes if n != node]
        g[node] = random.sample(candidates, k)
    return g

graph_low = {
    'Kim': ['Lee', 'Park', 'Choi'],
    'Lee': ['Jung', 'Kang'],
    'Park': ['Cho'],
    'Choi': ['Yoon', 'Jang'],
    'Jung': ['Lim'],
    'Kang': [],
    'Cho': ['Lim'],
    'Yoon': [],
    'Jang': [],
    'Lim': []
}

graph_mid_1 = generate_random_graph(num_nodes=100, max_edges=3, seed=42)

graph_mid_2 = generate_random_graph(num_nodes=100, max_edges=2, seed=77)

print("[테스트 케이스 1 - 난이도 하: 노드 10개, max_depth=2, 시작='Kim']")
trace_infection(graph_low, 'Kim', max_depth=2)

print("[테스트 케이스 2 - 난이도 중(1): 노드 100개, max_depth=10, 시작='Node_0']")
trace_infection(graph_mid_1, 'Node_0', max_depth=10)

print("[테스트 케이스 3 - 난이도 중(2): 노드 100개, max_depth=10, 시작='Node_35']")
trace_infection(graph_mid_2, 'Node_35', max_depth=10)
```
