```python
from collections import deque

graph = {
    '안우현': ['최우진', '배정우', '김서준', '이서아', '박도윤'],
    '이가을': ['김준석', '강석민', '김진우', '소준영', '김유진'],
    '박채완': ['이도현', '이대환', '오수아', '차재욱', '조형철'],
    '정민서': ['이준수', '김준성', '최하은', '정시우', '강유나'],
    '윤여주': ['김수인', '황정민', '김민정', '정다연', '안서윤'],
    '권태현': ['김주홍', '염도형', '김기헌', '박진성', '나희재']
}

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    while queue:
        v = queue.popleft()
        print(v, end=' ')
        for neighbor in graph.get(v, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

print("\nBFS 탐색 순서:")
bfs(graph, '안우현')
```

```python
from collections import deque

graph = {
    '안우현': ['최우진', '배정우', '김서준', '이서아', '박도윤'],
    '이가을': ['김준석', '강석민', '김진우', '소준영', '김유진'],
    '박채완': ['이도현', '이대환', '오수아', '차재욱', '조형철'],
    '정민서': ['이준수', '김준성', '최하은', '정시우', '강유나'],
    '윤여주': ['김수인', '황정민', '김민정', '정다연', '안서윤'],
    '권태현': ['김주홍', '염도형', '김기헌', '박진성', '나희재']
}

def recommend_friend(graph, start):
    visited = set()
    queue = deque([(start, 0)])
    print(f"\n{start}의 친구 추천 (2단계):")

    while queue:
        current, depth = queue.popleft()
        if current not in visited:
            visited.add(current)
            if depth == 2:
                print("-", current)
            if depth < 2:
                for neighbor in graph.get(current, []):
                    queue.append((neighbor, depth + 1))

recommend_friend(graph, '안우현')
```
